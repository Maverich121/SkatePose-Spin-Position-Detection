```python
movement_start_frame = 60
movement_end_frame = 145

data['is_movement'] = data['frame'].between(movement_start_frame, movement_end_frame)
```

```python
data
```

```python
# plot the movement from frame 60 to 145
import matplotlib.pyplot as plt
movement_data = data[data['is_movement']]
landmark_data = movement_data[movement_data['landmark'] == "RIGHT_ANKLE"]

plt.figure(figsize=(10, 6))

plt.plot(landmark_data['x'], landmark_data['y'], label=f'Right Ankle')
plt.gca().invert_yaxis()  # Invert y-axis for correct orientation
plt.title('Landmark Trajectories (Frames 60-145)')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()
plt.grid()
plt.show()
```
